```python
dept_counts = df["dept"].value_counts()
count_arr = dept_counts.values
count_arr
```

```python
"total"
total = np.sum(count_arr)
print("total:",total)
#percentage
percentage=(count_arr/total)*100
print("percentage:",percentage)
```

```python
df["timeStamp"] = pd.to_datetime(df["timeStamp"])
hour_arr =df["timeStamp"].dt.hour.to_numpy()
hour_arr
hour_count=np.bincount(hour_arr)
#printing the result in a cleaner format
for hour,count in enumerate(hour_count):
    print(f"{hour} hour:{count} calls")
```

```python
#for daytime 8:00 to 18:00
daytime = (hour_arr>= 8) &(hour_arr<=18)
nighttime = ~daytime

daytime_calls = np.sum(daytime)
nighttime_calls = np.sum(nighttime)
print("daytime calls:", daytime_calls)
print("nigthttime calls:",nighttime_calls)
```

C3-MATPLOTLIB:VISUALIZING THE 911 DATASET

```python
dept_counts= df["dept"].value_counts()
plt.figure(figsize=(10,6))
plt.bar(dept_counts.index,dept_counts)
plt.title("total number of calls")
plt.xlabel("department")
plt.ylabel("nuumber of calls")
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.hist(hour_arr,bins=24)
plt.title("distribution of 911 calls by hour of day")
plt.xlabel("hour of the day")
plt.ylabel("number of calls")
plt.show()
```

```python
twp = df["twp"].value_counts()
plt.figure(figsize=(10,7))
plt.barh(twp.index[::-1],twp.values[::-1])
plt.xlabel("number of calls")
plt.ylabel("Township")
plt.show()
```

```python
df["month"] = df["timeStamp"].dt.month
monthly = df.groupby("month")["e"].count()
plt.figure(figsize=(10,6))
plt.plot(monthly.index,monthly.values)
plt.title("total 911 calls per month")
plt.xlabel("month")
plt.ylabel("total number of calls")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df,x = "dept",palette="Set2")
plt.title("number of 911 calls per day")
plt.xlabel("deparment")
plt.ylabel("count")
plt.show()
```

```python
df["day_of_week"] = df["timeStamp"].dt.dayofweek
df["hour"] =df["timeStamp"].dt.hour
pivot =df.groupby(["day_of_week","hour"])["e"].count().unstack()
plt.figure(figsize=(14,6))
sns.heatmap(pivot)
plt.title("call volume by day and hour")
plt.xlabel("hour of the day")
plt.ylabel("day of week(0=monday,6=sunday")
plt.show()
```

```python
plt.figure(figsize=(9,6))
sns.boxplot(data=df,x="dept",y="lat",palette="Blues")
plt.title("latitude distribution across departments")
plt.xlabel("Department")
plt.ylabel("Latitude")
plt.show()
```

```python
ems_df =df[df["dept"] == "EMS"]
top5 = ems_df["reason"].value_counts().head(5).index
top = ems_df[ems_df["reason"].isin(top5)]

plt.figure(figsize=(12,6))
sns.countplot(data=top,x ="reason",palette="Set1")
plt.title("TOP 5 REASONS FOR EMS CALL")
plt.xlabel("Reason")
plt.ylabel("Count")
plt.show()
```

FROM THE HEATMAP:most of the emergency calls are made durig the  9-17 hours which is the timw most people will be at work  and wednesday,thrusday abd friday  shows the highes number of emergency calls during those hours and 0-6th hour and 18-19th  of the day shows the least amount of calls because most people are still on bed  by then  and teh days with the least amount of calls are saturday and sunday
